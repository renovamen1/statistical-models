# this notebook covers the implementation of Batch Gradient Descent.

```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_diabetes
from sklearn.metrics import r2_score
from sklearn.model_selection import train_test_split

```

```python
X,y = load_diabetes(return_X_y=True)
```

```python
print(X.shape)
print(y.shape)
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```


# Getting the actual parameter values using Sklearn

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(X_train,y_train)
y_pred = lr.predict(X_test)
```

```python
lr.intercept_
```

```python
r2_score(y_test,y_pred)
```

# Building the batch gradient descent class from scratch

```python
class BatchGD:
    def __init__(self,learning_rate=0.01,epochs=50):
        self.intercept = None
        self.lr = learning_rate
        self.epochs = epochs
        
    def fit(self,X_train,y_train):
        # initalizing the coefficients
        self.intercept = 0
        self.coeff = np.ones(X_train.shape[1])
        
        for i in range(self.epochs):
            # updating the coefficients and intercept
            y_pred = self.intercept + np.dot(X_train,self.coeff)
            slope_at_intercept = -2 * np.mean( y_train - y_pred)
            self.intercept = self.intercept - self.lr * slope_at_intercept

            slope_of_coeff = (-2  *  np.dot(y_train - y_pred,X_train))/ X_train.shape[0] 
            self.coeff = self.coeff - self.lr * slope_of_coeff



    def predict(self,X_test):
        return self.intercept + np.dot(X_test,self.coeff)
```

```python
GD = BatchGD(learning_rate=0.1,epochs=10000)
GD.fit(X_train,y_train)
print(GD.intercept)
print(GD.coeff)
```

```python
y_pred = GD.predict(X_test)
r2_score(y_test,y_pred)
```
