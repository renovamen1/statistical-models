# this notebook covers the implementation of Stochastic Gradient Descent.

```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_diabetes
from sklearn.metrics import r2_score
from sklearn.model_selection import train_test_split
import time
```

```python
X,y = load_diabetes(return_X_y=True)
```

```python
print(X.shape)
print(y.shape)
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```


# Getting the actual parameter values using Sklearn

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(X_train,y_train)
y_pred = lr.predict(X_test)
```

```python
print(lr.coef_)
lr.intercept_
```

```python
r2_score(y_test,y_pred)
```

# Building the Stochastic gradient descent class from scratch

```python
class SGDRegressor:
    def __init__(self,learning_rate=0.01,epochs=50):
        self.intercept = None
        self.lr = learning_rate
        self.epochs = epochs
        
    def fit(self,X_train,y_train):
        # initalizing the coefficients
        self.intercept = 0
        self.coeff = np.ones(X_train.shape[1])
        
        for i in range(self.epochs):
            
            for j in range(X_train.shape[0]):
                # we are randomly selecting a row to predict its y_hat/y_pred
                idx = np.random.randint(0,X_train.shape[0])
                y_pred = self.intercept + np.dot(X_train[idx],self.coeff)

                slope_at_intercept = -2  * (y_train[idx] - y_pred)
                self.intercept = self.intercept - self.lr * slope_at_intercept

                slope_of_coeffs = - 2 * np.dot((y_train[idx] - y_pred),X_train[idx])
                self.coeff = self.coeff - (self.lr * slope_of_coeffs)


    


    def predict(self,X_test):
        return self.intercept + np.dot(X_test,self.coeff)
```

```python
sgd = SGDRegressor(learning_rate=0.01,epochs=60)
start = time.time()
sgd.fit(X_train,y_train)
print("The time take is",time.time() - start)
print(sgd.intercept)
print(sgd.coeff)
```

```python
y_pred = sgd.predict(X_test)
r2_score(y_test,y_pred)
```
