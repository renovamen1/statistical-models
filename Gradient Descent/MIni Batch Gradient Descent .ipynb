# Building the Mini Batch gradient descent class from scratch

```python
class MBGDRegressor:
    def __init__(self,batch_size,learning_rate=0.01,epochs=50):
        self.intercept = None
        self.lr = learning_rate
        self.epochs = epochs
        self.batch_size = batch_size
        
    def fit(self,X_train,y_train):
        # initalizing the coefficients
        self.intercept = 0
        self.coeff = np.ones(X_train.shape[1])
        
        for i in range(self.epochs):
            
            for j in range(int(X_train.shape[0]/ self.batch_size)):
                # we are randomly selecting rows to predict  y_hat/y_pred of that batch
                idx = random.sample(range(X_train.shape[0]),self.batch_size)
                y_pred = self.intercept + np.dot(X_train[idx],self.coeff)

                slope_at_intercept = -2  * np.mean(y_train[idx] - y_pred)
                self.intercept = self.intercept - self.lr * slope_at_intercept

                slope_of_coeffs = - 2 * np.dot((y_train[idx] - y_pred),X_train[idx])
                self.coeff = self.coeff - (self.lr * slope_of_coeffs)

    def predict(self,X_test):
        return self.intercept + np.dot(X_test,self.coeff)
```

```python
mbgd = MBGDRegressor(batch_size = 10,learning_rate=0.01,epochs=50)

mbgd.fit(X_train,y_train)
print(mbgd.intercept)
print(mbgd.coeff)
```
