# this notebook includes the demo implementation of regresssion metrics like MAE,MSE,RMSE, R2 score,Adjusted R2 score and thier workings.

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('placement.csv')
```

```python
df.head()
df.shape
```

```python
plt.scatter(df['cgpa'],df['package'])
plt.xlabel('CGPA')
plt.ylabel('Package(in lpa)')
```

```python
X = df.iloc[:,0:1]
y = df.iloc[:,-1]
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
```

```python
lr.fit(X_train,y_train)
```

```python
plt.scatter(df['cgpa'],df['package'])
plt.plot(X_train,lr.predict(X_train),color='red')
plt.xlabel('CGPA')
plt.ylabel('Package(in lpa)')
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
```

```python
y_pred = lr.predict(X_test)
```

```python
y_test.values
```

```python
print("MAE",mean_absolute_error(y_test,y_pred))
```

```python
print("MSE",mean_squared_error(y_test,y_pred))
```

```python
print("RMSE",np.sqrt(mean_squared_error(y_test,y_pred)))
```

```python
print("R2 score",r2_score(y_test,y_pred))
r2 = r2_score(y_test,y_pred)
```

```python
# Adjusted R2 score
X_test.shape[0]
```

```python
Adjusted_R2 = 1 - ((1-r2)*(X_test.shape[0] - 1 )/(X_test.shape[0]-1-X_test.shape[1]))
Adjusted_R2
```

# Observation of R2 score When adding a irrelevant feature

```python
new_df1 = df.copy()
new_df1['random_feature'] = np.random.random(200)

new_df1 = new_df1[['cgpa','random_feature','package']]
new_df1.head()
```

```python
# making sure the random feature is completely random.
plt.scatter(new_df1['random_feature'],new_df1['package'])
plt.xlabel('random_feature')
plt.ylabel('Package(in lpa)')
```

```python
X = new_df1.iloc[:,0:2]
y = new_df1.iloc[:,-1]
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
lr = LinearRegression()
```

```python
lr.fit(X_train,y_train)
```

```python
y_pred = lr.predict(X_test)
```

# we can see R2 score still increases when adding a irrelevant feature.

```python
print("R2 score",r2_score(y_test,y_pred))
r2 = r2_score(y_test,y_pred)
```

# but adjusted R2 score decreases

```python
Adjusted_R2 = 1 - ((1-r2)*(X_test.shape[0] - 1 )/(X_test.shape[0]-1-X_test.shape[1]))
Adjusted_R2
```
