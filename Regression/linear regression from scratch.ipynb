# This notebook includes the implementation of simple regression model from scratch.

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("placement.csv")
df
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(df.iloc[:,0],df.iloc[:,-1],test_size=0.2,random_state=42)
```

```python
class myLR:
    def __init__(self):
        pass

    def fit(self,X,y):
        self.coef = ((X - X.mean()) * (y - y.mean())).sum() / ((X - X.mean())**2).sum()
        self.intercept = y.mean() - (self.coef * X.mean())

    def predict(self,X):
        return (self.coef * X) + self.intercept

        
```

```python
lr = myLR()
lr.fit(X_train,y_train)
```

```python
lr.coef
```

```python
lr.intercept
```

```python
y_pred = lr.predict(X_test)
```

```python
from sklearn.metrics import r2_score
r2_score(y_test,y_pred)
```
